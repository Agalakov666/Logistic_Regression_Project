### О проекте и бизнес-задаче

Компания Happy Beans Coffee запустила сервис доставки элитного кофе и столкнулась с проблемой высокого оттока клиентов — ежемесячно база теряет 10% пользователей. В условиях высокой конкуренции привлечение нового клиента обходится в 5–7 раз дороже, чем удержание существующего.

Цель проекта — построить модель бинарной классификации, которая спрогнозирует вероятность оттока каждого клиента в следующем месяце. Это позволит бизнесу:

- Точечно работать с клиентами из группы риска, предлагая персональные скидки и промокоды.

- Оптимизировать маркетинговый бюджет, не тратя его на лояльных пользователей, которые не собираются уходить.

- Повысить прогнозируемость будущей выручки (Revenue Forecasting).

### Данные

Проект основан на аналитической витрине coffee_churn_dataset, собранной из транзакционной базы, логов мобильного приложения и опросов клиентов.

- Структура: Каждая строка — это уникальный клиент. Признаки агрегированы за последние 4 недели.

- Признаки: Включают числовые метрики (суммы трат, частота заказов, средний чек, давность последнего заказа, количество открытий приложения), категориальные данные (тип подписки, предпочтения по кофе и молоку, тип устройства) и бинарные флаги (включены ли уведомления, пробовал ли сезонное меню).

- Целевая переменная: churn (перестал ли пользователь пользоваться сервисом).

### Методология и выбор метрик

Учитывая, что подавляющее большинство клиентов остаются (классы несбалансированы), классическая метрика Accuracy будет неинформативной. Основной метрикой выбрана Precision-Recall AUC (PR AUC), которая фокусируется на качестве предсказания меньшего, но ключевого для бизнеса класса — уходящих клиентов. Также будут отслеживаться Precision, Recall и F1-мера для поиска «золотой середины» между охватом и точностью.

Тип модели: Для обеспечения интерпретируемости результатов (бизнесу важно понимать, какие именно признаки влияют на отток) будет использоваться линейная модель LogisticRegression. В качестве точки отсчета (baseline) выступит DummyClassifier.

```python
correlations_with_target = df[num_cols+['churn']].phik_matrix()[['churn']].sort_values(by='churn', ascending=False)
print(correlations_with_target)
plt.figure(figsize=(2, 10))
sns.heatmap(correlations_with_target, annot=True, fmt='.2f', cmap='coolwarm', vmin=0, vmax=1)
plt.title('Корреляционная матрица Phi_K (только значимые связи)')
plt.show()
```

```python
def check_collinearity(df, threshold=0.8, target_col=None):
    """
    Находит сильно коррелирующие пары признаков
    threshold - порог корреляции 
    """
    if target_col and target_col in df.columns:
        X = df.drop(columns=[target_col])
    else:
        X = df.copy()
    
    corr_matrix = X.corr()
    
    high_corr = []
    for i in range(len(corr_matrix.columns)):
        for j in range(i+1, len(corr_matrix.columns)):
            if abs(corr_matrix.iloc[i, j]) > threshold:
                high_corr.append({
                    'feature1': corr_matrix.columns[i],
                    'feature2': corr_matrix.columns[j],
                    'correlation': corr_matrix.iloc[i, j]
                })
    
    return pd.DataFrame(high_corr).sort_values('correlation', ascending=False)

collinear_pairs = check_collinearity(df[num_cols+['churn']], threshold=0.8, target_col='weight')
print(collinear_pairs)
```

```python
def calculate_improvement(baseline_results, improved_results, baseline_name="Baseline", improved_name="Improved"):
    """
    Считает разницу метрик между двумя моделями.
    """
    metrics = ['test_precision', 'test_recall', 'test_f1', 'test_average_precision']
    comparison_data = []

    for metric in metrics:
        base_mean = np.mean(baseline_results[metric])
        imp_mean = np.mean(improved_results[metric])
        diff = imp_mean - base_mean
        
        # Считаем процент улучшения (защита от деления на ноль)
        percent_diff = (diff / base_mean * 100) if base_mean != 0 else np.nan
        
        comparison_data.append({
            'Metric': metric.replace('test_', '').replace('_', ' ').title(),
            f'{baseline_name}': round(base_mean, 3),
            f'{improved_name}': round(imp_mean, 3),
            'Improvement (abs)': round(diff, 3),
            'Improvement (%)': f"{percent_diff:+.1f}%" if not np.isnan(percent_diff) else "N/A"
        })

    df = pd.DataFrame(comparison_data).set_index('Metric')
    return df
improvement_df = calculate_improvement(
    log_model, 
    log_model_fe
)
display(improvement_df)

```

### Составляем таблицу

```python
def format_cv_results(cv_results, metric='mean_test_score'):
    """
    Создает датафрейм из результатов GridSearchCV, извлекает гиперпараметры,
    добавляет метрику и сортирует по убыванию метрики.
    """
    # 1. Извлекаем словари с гиперпараметрами
    params = cv_results['params']
    
    # 2. Преобразуем список словарей в DataFrame
    df = pd.DataFrame(params)
    
    # 3. Добавляем колонку с нашей ключевой метрикой
    df['average_precision'] = cv_results[metric]
    
    # 4. Сортируем по метрике (от лучшего к худшему) и сбрасываем индексы
    df = df.sort_values(by='average_precision', ascending=False).reset_index(drop=True)
    
    return df

results_df = format_cv_results(grid_search.cv_results_)
```

```python
def evaluate_on_test(y_true, y_pred, y_proba):
    """
    Считает финальные метрики на тестовой выборке и возвращает DataFrame.
    """
    metrics = {
        'Precision': precision_score(y_true, y_pred),
        'Recall': recall_score(y_true, y_pred),
        'F1-Score': f1_score(y_true, y_pred),
        'Average Precision': average_precision_score(y_true, y_proba)
    }
    
    df = pd.DataFrame(list(metrics.items()), columns=['Metric', 'Test Score'])
    df['Test Score'] = df['Test Score'].round(3) # Округляем для красоты
    
    return df
```
